```python
import pandas as pd
import json

# Load Suicide data
with open("/content/drive/MyDrive/BUSAN300_Project_Rport/Suicide.json", "r", encoding="utf-8") as f:
    suicide_data = json.load(f)

# Extract suicide rates
suicide_entries = []
for country in suicide_data['dimension'][3]['code']:  # COUNTRY
    country_code = country['label']
    country_name = country['display']
    suicide_attrs = {attr['category']: attr['value'] for attr in country.get('attr', [])}

    # Only add country data if suicide rate info is available
    for year_info in suicide_data['dimension'][2]['code']:  # YEAR
        year = year_info['label']
        suicide_entries.append({
            "Country_Code": country_code,
            "Country": country_name,
            "Year": int(year),
            "WHO_Region": suicide_attrs.get("WHO_REGION"),
            "Income_Level": suicide_attrs.get("WORLD_BANK_INCOME_GROUP")
            # Suicide rate will be merged in later
        })

suicide_df = pd.DataFrame(suicide_entries)

# NOTE: The actual suicide rates would normally be under another field (like "value" or "dataset"),
# but since the rate data wasn't visible in the JSON, assume preformatted 'suicide_df' needs external rates added or matched later.

# Load GDP data
gdp_df = pd.read_csv("/content/drive/MyDrive/BUSAN300_Project_Rport/GDP.csv")

# Basic GDP cleaning
print("Columns in GDP_df before renaming:", gdp_df.columns.tolist())
gdp_df = gdp_df.rename(columns={"Country Code": "Country_Code", "Year": "Year", "Value": "GDP"})
gdp_df = gdp_df[["Country_Code", "Year", "GDP"]]
gdp_df['Year'] = gdp_df['Year'].astype(int)

# Merge GDP with suicide data
combined_df = pd.merge(suicide_df, gdp_df, on=["Country_Code", "Year"], how="inner")

# Export to Excel
combined_df.to_excel("Combined_Suicide_GDP.xlsx", index=False)

print("Combined dataset exported to 'Combined_Suicide_GDP.xlsx'.")
```
